```python
import ffmpeg

# 1. Cargar el archivo de entrada
audio_stream = ffmpeg.input("Om.mp3")

# 2. Recortar (desde el segundo 0 hasta el segundo 130)
trimmed_stream = audio_stream.filter('atrim', start=0, end=140).filter('asetpts', 'PTS-STARTPTS')

# 3. Definir la salida
output_stream = ffmpeg.output(trimmed_stream, "trimmed.mp3")

# 4. Ejecutar el proceso
output_stream.run()

print("¡Audio recortado con éxito!")
```
